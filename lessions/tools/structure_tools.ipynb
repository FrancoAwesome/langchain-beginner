## Structure Tools
Define two functions for number multiple
one for sync and another for async

```python
from langchain_core.tools import StructuredTool

def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    print(f"sync -> a: {a}, b: {b}")
    return a * b

async def amultiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    print(f"async -> a: {a}, b: {b}")
    return a * b


# func --> sync func, invoke func in sync context
# coroutine --> async func, invoke func in async context
calculator = StructuredTool.from_function(
    func=multiply,
    coroutine=amultiply,
    return_result=True
)
# sync -> a:1, b:2
print(calculator.invoke({"a":1,"b":2}))
# async -> a:2, b:5
print(await calculator.ainvoke({"a":2,"b":5}))
```

## Structure Tools with args configuration
Use arg configuration for params description in the funtions

```python
from pydantic import BaseModel, Field

class CalculateInput(BaseModel):
    a: int = Field(description="first number")
    b: int = Field(description="second number")

# func --> sync func, invoke func in sync context
# coroutine --> async func, invoke func in async context
calculator = StructuredTool.from_function(
    func=multiply,
    coroutine=amultiply,
    name="multiply",
    description="multiply two numbers",
    args=CalculateInput,
    return_direct=True
)
# sync -> a:1, b:2
print(calculator.invoke({"a":1,"b":2}))
# async -> a:2, b:5
print(await calculator.ainvoke({"a":2,"b":5}))
```
